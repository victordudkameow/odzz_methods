```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# Настройка стиля графиков
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
plt.rcParams['font.size'] = 10
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['figure.figsize'] = (12, 6)

# =====================================================
# 0. НАСТРОЙКИ ПО УМОЛЧАНИЮ
# =====================================================

DEFAULT_CLASSES = {
    0: 'Вода',
    1: 'Древесно-кустарниковая растительность',
    2: 'Зоны засоления с галофитной растительностью',
    3: 'Пустыня',
    4: 'Зоны осушения воды'
}

DEFAULT_START_YEAR = 2006

# =====================================================
# 1. ПОЛУЧЕНИЕ НАСТРОЕК ОТ ПОЛЬЗОВАТЕЛЯ
# =====================================================

def get_user_settings():
    """Запрашивает у пользователя параметры анализа"""
    print("\n" + "="*80)
    print("⚙️  НАСТРОЙКА ПАРАМЕТРОВ АНАЛИЗА (ПО ГОДАМ)")
    print("="*80)

    # 1. Имя файла
    filepath = input(f"\n📂 Введите путь к файлу (Enter для 'data/tableofclasses.csv'): ").strip()
    if not filepath:
        filepath = 'data/tableofclasses.csv'
    
    # 2. Начальный год
    try:
        start_year = input(f"\n📅 Начальный год (Enter для {DEFAULT_START_YEAR}): ").strip()
        start_year = int(start_year) if start_year else DEFAULT_START_YEAR
    except ValueError:
        print("⚠️  Ошибка ввода года. Используем значение по умолчанию.")
        start_year = DEFAULT_START_YEAR

    # 3. Год окончания
    try:
        end_year = input(f"📅 Год окончания анализа (Enter для автоопределения): ").strip()
        end_year = int(end_year) if end_year else None
    except ValueError:
        print("⚠️  Ошибка ввода года. Будет определено автоматически.")
        end_year = None

    # 4. Количество классов
    print(f"\n🏷️  ВВОД НАЗВАНИЙ КЛАССОВ")
    print(f"   (Нажмите Enter, чтобы оставить значение по умолчанию)")
    print("-"*80)
    
    class_names = {}
    num_classes_input = input(f"\n   Количество классов (Enter для {len(DEFAULT_CLASSES)}): ").strip()
    num_classes = int(num_classes_input) if num_classes_input else len(DEFAULT_CLASSES)
    
    for i in range(num_classes):
        default_name = DEFAULT_CLASSES.get(i, f'Класс {i}')
        class_input = input(f"   Класс {i} [{default_name}]: ").strip()
        if class_input:
            class_names[i] = class_input
        else:
            class_names[i] = default_name
    
    print(f"\n✅ Введено классов: {len(class_names)}")
    for cls_id, cls_name in class_names.items():
        print(f"     {cls_id} - {cls_name}")

    return filepath, start_year, end_year, class_names, num_classes

# =====================================================
# 2. ЗАГРУЗКА ДАННЫХ И ОПРЕДЕЛЕНИЕ ПЕРИОДОВ
# =====================================================

def load_data_and_detect_periods(filepath, start_year, end_year):
    """Загружает CSV и определяет количество лет на основе года окончания"""
    
    print(f"\n📂 Загрузка файла: {filepath}")
    
    if not Path(filepath).exists():
        print(f"❌ Файл не найден: {filepath}")
        return None, None, None
    
    df = pd.read_csv(
        filepath, 
        sep=';', 
        header=0,
        decimal=',',
        thousands=' ',
        encoding='utf-8',
        dtype={'OID_': str, 'trajectory': str, 'AREA': float}
    )
    
    df.columns = ['id', 'trajectory', 'area']
    print(f"✅ Загружено {len(df):,} записей")
    
    # Очищаем траектории
    def clean_code(code):
        if pd.isna(code):
            return ''
        code = str(code).strip()
        if 'E' in code.upper():
            try:
                code_clean = code.replace(',', '.')
                num = float(code_clean)
                code = str(int(num))
            except:
                return ''
        # Оставляем только цифры
        code = ''.join(c for c in code if c.isdigit())
        return code
    
    df['trajectory_clean'] = df['trajectory'].apply(clean_code)
    
    # Определяем максимальную длину траектории
    valid_trajectories = df[df['trajectory_clean'].str.len() > 0]
    
    if len(valid_trajectories) == 0:
        print("❌ Нет валидных траекторий для анализа")
        return None, None, None
    
    # Берём наиболее частую длину траектории (мода)
    length_counts = valid_trajectories['trajectory_clean'].str.len().value_counts()
    max_length = length_counts.index[0]
    
    print(f"📅 Обнаружена длина траектории: {max_length} символов")
    print(f"   Валидных записей: {len(valid_trajectories)} из {len(df)}")
    
    # Определяем количество лет для анализа
    if end_year is not None:
        num_years = end_year - start_year + 1
        print(f"\n🎯 Пользователь задал год окончания: {end_year}")
        print(f"   Период анализа: {start_year} - {end_year} ({num_years} лет)")
        
        if num_years > max_length:
            print(f"⚠️  Предупреждение: запрошено {num_years} лет, но максимальная длина траектории {max_length}")
            print(f"   Будет использовано {max_length} лет")
            num_years = max_length
    else:
        num_years = max_length
        detected_end_year = start_year + num_years - 1
        print(f"\n🎯 Год окончания определен автоматически: {detected_end_year}")
        print(f"   Период анализа: {start_year} - {detected_end_year} ({num_years} лет)")
    
    # 🔥 Нормализация: обрезаем траектории до нужной длины (с начала!)
    print(f"\n🔧 Нормализация траекторий (обрезка до {num_years} символов с начала)...")
    
    def normalize_trajectory(traj, expected_length):
        if pd.isna(traj) or traj == '':
            return '0' * expected_length  # Пустая → все нули
        if len(traj) < expected_length:
            return traj.zfill(expected_length)  # Короткая → дополняем нулями слева
        elif len(traj) > expected_length:
            return traj[:expected_length]  # 🔥 Длинная → берем ПЕРВЫЕ N символов (с начала!)
        else:
            return traj  # Нужная длина → оставляем как есть
    
    df['trajectory_clean'] = df['trajectory_clean'].apply(lambda x: normalize_trajectory(x, num_years))
    
    # Проверка после нормализации
    invalid_after = df[df['trajectory_clean'].str.len() != num_years]
    if len(invalid_after) > 0:
        print(f"⚠️  Предупреждение: {len(invalid_after)} записей всё ещё имеют некорректную длину")
    else:
        print(f"✅ Все траектории нормализованы до {num_years} символов")
    
    # Примеры траекторий для проверки
    print(f"\n📋 Примеры траекторий после нормализации:")
    for i in range(min(5, len(df))):
        print(f"   {df.iloc[i]['id']}: {df.iloc[i]['trajectory_clean']}")
    
    print(f"\n📊 Площадь: мин={df['area'].min():,.2f}, макс={df['area'].max():,.2f}, среднее={df['area'].mean():,.2f}")
    
    return df, num_years, num_years

# =====================================================
# 3. ИЗВЛЕЧЕНИЕ КЛАССОВ ПО ГОДАМ (каждая позиция = год)
# =====================================================

def extract_yearly_classes(df, start_year, num_years):
    """
    Добавляет столбцы с классом для каждого года.
    Каждая позиция в траектории = один год (уже готовое значение).
    Траектории уже нормализованы (дополнены нулями слева).
    """
    print(f"\n🔄 Извлечение классов по годам (каждая позиция = год)...")
    
    for year_idx in range(num_years):
        year = start_year + year_idx
        col_name = f'class_{year}'
        
        # Извлекаем символ на позиции year_idx
        def get_class_at_position(traj, pos):
            if pd.isna(traj) or len(traj) <= pos:
                return 0
            try:
                return int(traj[pos])
            except (ValueError, IndexError):
                return 0
        
        df[col_name] = df['trajectory_clean'].apply(lambda x: get_class_at_position(x, year_idx))
    
    print(f"✅ Извлечено {num_years} годовых колонок ({start_year} - {start_year + num_years - 1})")
    return df

# =====================================================
# 4. СТАТИСТИКА ПО ГОДАМ
# =====================================================

def calculate_yearly_statistics(df, start_year, num_years, class_names):
    """Считает статистику по каждому году"""
    
    print("📊 Расчёт статистики по годам...")
    
    results = []
    total_pixels = len(df)
    total_area = df['area'].sum()
    num_classes = len(class_names)
    
    for year_idx in range(num_years):
        year = start_year + year_idx
        col_name = f'class_{year}'
        
        if col_name not in df.columns:
            continue
            
        class_counts = df[col_name].value_counts().sort_index()
        class_areas = df.groupby(col_name)['area'].sum()
        
        for cls in range(num_classes):
            count = class_counts.get(cls, 0)
            area = class_areas.get(cls, 0) if not class_areas.empty else 0
            pct_pixels = (count / total_pixels * 100) if total_pixels > 0 else 0
            pct_area = (area / total_area * 100) if total_area > 0 else 0
            
            results.append({
                'year': year,
                'class_id': cls,
                'class_name': class_names.get(cls, f'Class {cls}'),
                'pixel_count': int(count),
                'pixel_percent': round(pct_pixels, 2),
                'area_sum': round(area, 2),
                'area_percent': round(pct_area, 2)
            })
    
    return pd.DataFrame(results)

# =====================================================
# 5. АНАЛИЗ ИЗМЕНЕНИЙ ПО ГОДАМ
# =====================================================

def analyze_yearly_transitions(df, start_year, num_years, class_names):
    """Анализирует переходы между классами от года к году"""
    
    print("🔄 Анализ переходов между годами...")
    
    transitions = []
    num_classes = len(class_names)
    
    for year_idx in range(num_years - 1):
        year_from = start_year + year_idx
        year_to = start_year + year_idx + 1
        
        col_from = f'class_{year_from}'
        col_to = f'class_{year_to}'
        
        if col_from not in df.columns or col_to not in df.columns:
            continue
        
        for from_cls in range(num_classes):
            for to_cls in range(num_classes):
                if from_cls != to_cls:
                    mask = (df[col_from] == from_cls) & (df[col_to] == to_cls)
                    count = mask.sum()
                    area = df[mask]['area'].sum() if count > 0 else 0
                    
                    if count > 0:
                        transitions.append({
                            'from_year': year_from,
                            'to_year': year_to,
                            'from_class_id': from_cls,
                            'to_class_id': to_cls,
                            'from_class': class_names.get(from_cls, f'Class {from_cls}'),
                            'to_class': class_names.get(to_cls, f'Class {to_cls}'),
                            'pixel_count': int(count),
                            'area_sum': round(area, 2),
                            'transition': f'{from_cls}→{to_cls}'
                        })
    
    return pd.DataFrame(transitions).sort_values('pixel_count', ascending=False) if transitions else pd.DataFrame()

# =====================================================
# 6. ПОСТРОЕНИЕ ГРАФИКОВ
# =====================================================

def plot_yearly_distribution(yearly_stats, class_names, output_folder='data/plots'):
    """Строит столбчатые диаграммы распределения классов по годам"""
    Path(output_folder).mkdir(parents=True, exist_ok=True)
    
    plot_data = yearly_stats.pivot_table(
        index='year', 
        columns='class_name', 
        values='area_percent',
        fill_value=0
    )
    
    ax = plot_data.plot(kind='bar', stacked=True, figsize=(14, 7), colormap='tab20')
    plt.title('Распределение площади по классам земного покрова (по годам)', fontsize=14, fontweight='bold')
    plt.xlabel('Год', fontsize=11)
    plt.ylabel('Доля площади, %', fontsize=11)
    plt.legend(title='Класс', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=8)
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(f'{output_folder}/area_distribution_by_year.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"📈 Сохранён график: {output_folder}/area_distribution_by_year.png")
    
    for cls_name in class_names.values():
        if cls_name in plot_data.columns:
            plt.figure(figsize=(10, 5))
            plot_data[cls_name].plot(marker='o', linewidth=2, markersize=6)
            plt.title(f'Динамика доли площади: {cls_name}', fontsize=13, fontweight='bold')
            plt.xlabel('Год')
            plt.ylabel('Доля площади, %')
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            safe_name = "".join(c for c in cls_name if c.isalnum() or c in ' _-')
            plt.savefig(f'{output_folder}/dynamics_yearly_{safe_name}.png', dpi=300, bbox_inches='tight')
            plt.close()
    
    print(f"📈 Сохранены графики годовой динамики для каждого класса")


def plot_yearly_area_trends(yearly_stats, class_names, output_folder='data/plots'):
    """Строит графики трендов площади по годам"""
    Path(output_folder).mkdir(parents=True, exist_ok=True)
    
    area_pivot = yearly_stats.pivot_table(
        index='year', 
        columns='class_name', 
        values='area_sum',
        fill_value=0
    )
    
    ax = area_pivot.plot(kind='line', marker='o', figsize=(12, 6), colormap='tab10', linewidth=2)
    plt.title('Годовая динамика площади по классам', fontsize=14, fontweight='bold')
    plt.xlabel('Год')
    plt.ylabel('Площадь')
    plt.legend(title='Класс', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=9)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f'{output_folder}/yearly_area_trends.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"📈 Сохранён график: {output_folder}/yearly_area_trends.png")


def plot_yearly_transition_matrices(yearly_transitions, start_year, num_years, class_names, output_folder='data/plots'):
    """Строит отдельную матрицу переходов для каждой пары лет"""
    
    if len(yearly_transitions) == 0:
        print("⚠️  Нет данных для построения матриц переходов")
        return
    
    Path(output_folder).mkdir(parents=True, exist_ok=True)
    
    transition_pairs = yearly_transitions[['from_year', 'to_year']].drop_duplicates()
    
    num_classes = len(class_names)
    class_ids = list(range(num_classes))
    class_labels = [class_names.get(i, f'Class {i}') for i in class_ids]
    
    print(f"🔄 Строим матрицы переходов для {len(transition_pairs)} пар лет...")
    
    for _, row in transition_pairs.iterrows():
        year_from = int(row['from_year'])
        year_to = int(row['to_year'])
        
        pair_data = yearly_transitions[
            (yearly_transitions['from_year'] == year_from) & 
            (yearly_transitions['to_year'] == year_to)
        ]
        
        if len(pair_data) == 0:
            continue
        
        matrix = np.zeros((num_classes, num_classes))
        
        for _, t_row in pair_data.iterrows():
            from_cls = t_row['from_class_id']
            to_cls = t_row['to_class_id']
            matrix[from_cls, to_cls] = t_row['pixel_count']
        
        # Вариант 1: Абсолютные значения
        plt.figure(figsize=(9, 7))
        sns.heatmap(matrix, annot=True, fmt='.0f', cmap='Blues', 
                    xticklabels=class_labels, yticklabels=class_labels,
                    cbar_kws={'label': 'Количество пикселей'})
        plt.title(f'Переходы: {year_from} → {year_to}\n(абсолютные значения)', 
                  fontsize=12, fontweight='bold', pad=15)
        plt.xlabel('Класс в году назначения', fontsize=10)
        plt.ylabel('Класс в исходном году', fontsize=10)
        plt.xticks(rotation=45, ha='right', fontsize=8)
        plt.yticks(rotation=0, fontsize=8)
        plt.tight_layout()
        
        safe_name = f"{year_from}_to_{year_to}"
        plt.savefig(f'{output_folder}/transition_yearly_abs_{safe_name}.png', dpi=300, bbox_inches='tight')
        plt.close()
        
        # Вариант 2: Проценты
        row_sums = matrix.sum(axis=1, keepdims=True)
        matrix_pct = np.divide(matrix, row_sums, where=row_sums!=0) * 100
        
        plt.figure(figsize=(9, 7))
        sns.heatmap(matrix_pct, annot=True, fmt='.1f', cmap='YlOrRd',
                    xticklabels=class_labels, yticklabels=class_labels,
                    cbar_kws={'label': '% от исходного класса'})
        plt.title(f'Переходы: {year_from} → {year_to}\n(в % от исходного класса)', 
                  fontsize=12, fontweight='bold', pad=15)
        plt.xlabel('Класс в году назначения', fontsize=10)
        plt.ylabel('Класс в исходном году', fontsize=10)
        plt.xticks(rotation=45, ha='right', fontsize=8)
        plt.yticks(rotation=0, fontsize=8)
        plt.tight_layout()
        plt.savefig(f'{output_folder}/transition_yearly_pct_{safe_name}.png', dpi=300, bbox_inches='tight')
        plt.close()
        
        print(f"  ✓ {year_from} → {year_to}: 2 графика")
    
    print(f"✅ Всего сохранено {len(transition_pairs) * 2} матриц переходов")


def plot_yearly_transition_summary(yearly_transitions, start_year, num_years, class_names, output_folder='data/plots', max_cols=4):
    """Строит сетку мини-матриц для быстрого обзора всех переходов"""
    
    if len(yearly_transitions) == 0:
        return
    
    Path(output_folder).mkdir(parents=True, exist_ok=True)
    
    transition_pairs = yearly_transitions[['from_year', 'to_year']].drop_duplicates()
    n_pairs = len(transition_pairs)
    
    if n_pairs == 0:
        return
    
    n_cols = min(max_cols, n_pairs)
    n_rows = (n_pairs + n_cols - 1) // n_cols
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(5*n_cols, 4*n_rows), 
                            constrained_layout=True)
    if n_pairs == 1:
        axes = np.array([[axes]])
    elif n_rows == 1:
        axes = axes.reshape(1, -1)
    elif n_cols == 1:
        axes = axes.reshape(-1, 1)
    
    num_classes = len(class_names)
    
    for idx, (_, row) in enumerate(transition_pairs.iterrows()):
        year_from = int(row['from_year'])
        year_to = int(row['to_year'])
        
        pair_data = yearly_transitions[
            (yearly_transitions['from_year'] == year_from) & 
            (yearly_transitions['to_year'] == year_to)
        ]
        
        matrix = np.zeros((num_classes, num_classes))
        for _, t_row in pair_data.iterrows():
            matrix[t_row['from_class_id'], t_row['to_class_id']] = t_row['pixel_count']
        
        row_sums = matrix.sum(axis=1, keepdims=True)
        matrix_pct = np.divide(matrix, row_sums, where=row_sums!=0) * 100
        
        ax = axes[idx // n_cols, idx % n_cols]
        
        im = ax.imshow(matrix_pct, cmap='YlOrRd', aspect='auto', vmin=0, vmax=100)
        ax.set_title(f'{year_from}→{year_to}', fontsize=9, fontweight='bold')
        ax.set_xlabel('To', fontsize=7)
        ax.set_ylabel('From', fontsize=7)
        
        if num_classes <= 5:
            ax.set_xticks(np.arange(num_classes))
            ax.set_yticks(np.arange(num_classes))
            ax.set_xticklabels([class_names.get(i, str(i)) for i in range(num_classes)], 
                              rotation=45, ha='right', fontsize=6)
            ax.set_yticklabels([class_names.get(i, str(i)) for i in range(num_classes)], 
                              fontsize=6)
        
        for i in range(num_classes):
            for j in range(num_classes):
                if matrix_pct[i, j] > 5:
                    ax.text(j, i, f'{matrix_pct[i, j]:.0f}%', 
                           ha="center", va="center", color="black", fontsize=5)
    
    for idx in range(n_pairs, n_rows * n_cols):
        ax = axes[idx // n_cols, idx % n_cols]
        ax.axis('off')
    
    cbar_ax = fig.add_axes([0.92, 0.15, 0.02, 0.7])
    cbar = fig.colorbar(plt.cm.ScalarMappable(cmap='YlOrRd'), cax=cbar_ax)
    cbar.set_label('% переходов', rotation=270, labelpad=15)
    
    plt.suptitle('Сводная матрица переходов по годам', 
                 fontsize=14, fontweight='bold', y=1.02)
    plt.savefig(f'{output_folder}/transition_yearly_summary.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"📈 Сохранена сводная сетка переходов")

# =====================================================
# 7. ВЫВОД ОТЧЁТА
# =====================================================

def print_yearly_summary(df, yearly_stats, yearly_transitions, start_year, num_years, class_names):
    """Выводит сводный отчёт по годам"""
    
    print("\n" + "="*80)
    print("📊 СТАТИСТИКА ПО КЛАССАМ И ГОДАМ")
    print("="*80)
    
    print("\n📅 ПО ГОДАМ:")
    print("-"*80)
    
    for year_idx in range(num_years):
        year = start_year + year_idx
        year_data = yearly_stats[yearly_stats['year'] == year]
        
        print(f"\n{year}:")
        
        for _, row in year_data.iterrows():
            print(f"  {row['class_name']:50s} {row['pixel_count']:6d} пикс. "
                  f"({row['pixel_percent']:6.2f}%) | Площадь: {row['area_sum']:>15,.2f} "
                  f"({row['area_percent']:6.2f}%)")
    
    if len(yearly_transitions) > 0:
        print("\n\n🔄 ТОП-20 ИЗМЕНЕНИЙ КЛАССОВ (ПО ГОДАМ):")
        print("-"*80)
        for i, row in yearly_transitions.head(20).iterrows():
            print(f"  {row['from_year']}→{row['to_year']}: "
                  f"{row['from_class'][:25]:25s} → {row['to_class'][:25]:25s} | "
                  f"{row['pixel_count']:6d} пикс. | {row['area_sum']:>15,.2f}")

# =====================================================
# 8. АНАЛИЗ СУДЬБЫ КЛАССОВ (для всех классов)
# =====================================================

def analyze_class_fate_all(df, start_year, num_years, class_names):
    """
    Анализирует судьбу каждого класса: 
    куда перешли пиксели из начального года в конечный
    """
    print("\n" + "="*80)
    print("🔄 СУДЬБА КЛАССОВ: трансформация за период анализа")
    print("="*80)
    
    first_year = start_year
    last_year = start_year + num_years - 1
    num_classes = len(class_names)
    
    for cls_id in range(num_classes):
        cls_name = class_names.get(cls_id, f'Class {cls_id}')
        col_start = f'class_{first_year}'
        col_end = f'class_{last_year}'
        
        # Выбираем пиксели, которые были в этом классе в начальном году
        source_pixels = df[df[col_start] == cls_id]
        total_count = len(source_pixels)
        total_area = source_pixels['area'].sum()
        
        if total_count == 0:
            print(f"\n⚪ {cls_name} ({cls_id}): нет пикселей в {first_year} году")
            continue
        
        print(f"\n📊 {cls_name.upper()} (класс {cls_id}) — исходное состояние в {first_year}:")
        print(f"   Всего пикселей: {total_count:,} | Площадь: {total_area:,.2f}")
        print(f"   {'→ Перешли в класс:':<40} {'Пикс.':>8} {'%':>7} {'Площадь':>15}")
        print(f"   {'-'*75}")
        
        # Анализируем, куда перешли эти пиксели к последнему году
        for target_cls in range(num_classes):
            target_pixels = source_pixels[source_pixels[col_end] == target_cls]
            count = len(target_pixels)
            area = target_pixels['area'].sum()
            pct = (count / total_count * 100) if total_count > 0 else 0
            
            if count > 0:  # Показываем только значимые переходы
                marker = "✓" if target_cls == cls_id else "→"
                print(f"   {marker} {class_names.get(target_cls, f'Class {target_cls}'):<35s} "
                      f"{count:8,d} {pct:6.2f}% {area:15,.2f}")
        
        # Краткий итог
        stayed = len(source_pixels[source_pixels[col_end] == cls_id])
        stayed_pct = stayed / total_count * 100
        changed = total_count - stayed
        changed_pct = 100 - stayed_pct
        
        print(f"\n   📈 Итог: {stayed_pct:5.1f}% остались в классе | {changed_pct:5.1f}% изменились")

# =====================================================
# 9. ЭКСПОРТ РЕЗУЛЬТАТОВ
# =====================================================

def export_results(yearly_stats, yearly_transitions, output_folder='data'):
    """Экспортирует результаты в CSV-файлы"""
    
    Path(output_folder).mkdir(exist_ok=True)
    
    yearly_stats.to_csv(f'{output_folder}/analysis_by_year.csv', 
                       index=False, sep=';', encoding='utf-8-sig')
    
    if len(yearly_transitions) > 0:
        yearly_transitions.to_csv(f'{output_folder}/analysis_transitions_yearly.csv', 
                          index=False, sep=';', encoding='utf-8-sig')
    
    print(f"\n✅ Результаты экспортированы в папку '{output_folder}':")
    print(f"   • analysis_by_year.csv")
    print(f"   • analysis_transitions_yearly.csv")

# =====================================================
# 10. ОСНОВНАЯ ФУНКЦИЯ
# =====================================================

def main():
    """Запускает полный анализ по годам"""
    
    # 1. Получение настроек
    filepath, start_year, end_year, class_names, num_classes = get_user_settings()
    
    # 2. Загрузка и определение количества лет
    df, num_years, _ = load_data_and_detect_periods(filepath, start_year, end_year)
    if df is None:
        return
    
    print("\n" + "="*80)
    print(f"🚀 АНАЛИЗ ТРАЕКТОРИЙ ПО ГОДАМ ({start_year} - {start_year + num_years - 1})")
    print(f"   Количество лет: {num_years}")
    print("="*80 + "\n")
    
    # 3. Извлечение классов по годам (каждая позиция = год)
    df = extract_yearly_classes(df, start_year, num_years)
    
    # 4. Расчёт статистики по годам
    yearly_stats = calculate_yearly_statistics(df, start_year, num_years, class_names)
    yearly_transitions = analyze_yearly_transitions(df, start_year, num_years, class_names)
    
    # 5. Вывод отчёта
    print_yearly_summary(df, yearly_stats, yearly_transitions, start_year, num_years, class_names)
    
    # 6. 🔥 АНАЛИЗ СУДЬБЫ ВСЕХ КЛАССОВ
    analyze_class_fate_all(df, start_year, num_years, class_names)
    
    # 7. ПОСТРОЕНИЕ ГРАФИКОВ
    print("\n" + "="*80)
    print("📊 ПОСТРОЕНИЕ ГРАФИКОВ (ПО ГОДАМ)")
    print("="*80)
    try:
        plot_yearly_distribution(yearly_stats, class_names)
        plot_yearly_area_trends(yearly_stats, class_names)
        plot_yearly_transition_matrices(yearly_transitions, start_year, num_years, class_names)
        plot_yearly_transition_summary(yearly_transitions, start_year, num_years, class_names)
        print("✅ Все графики успешно построены!")
    except Exception as e:
        print(f"⚠️  Ошибка при построении графиков: {e}")
        import traceback
        traceback.print_exc()
        print("💡 Убедитесь, что установлены matplotlib и seaborn: pip install matplotlib seaborn")
    
    # 8. Экспорт
    export_results(yearly_stats, yearly_transitions)
    
    print("\n" + "="*80)
    print("✅ ✅ ✅ АНАЛИЗ ПО ГОДАМ ЗАВЕРШЕН ✅ ✅ ✅")
    print("="*80 + "\n")
    
    return df, yearly_stats, yearly_transitions

# =====================================================
# ЗАПУСК
# =====================================================

if __name__ == "__main__":
    df, yearly_stats, yearly_transitions = main()
```
